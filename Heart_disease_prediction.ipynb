**Import all the needed libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

**Data Collection**

```python
# import the dataset to pandas

df = pd.read_csv(r"C:\Users\Admin\Jupyter_Note_folder\Data Science Projects\data\heart_disease_data.csv")

df.head()
```

```python
# reshuffle the dataset

df = df.sample(frac = 1, random_state = 42).reset_index(drop = True)

df.head()
```

```python
# shape of the data

df.shape
```

```python
# information of the dataset

df.info()
```

```python
# check for missing value

df.isnull().sum()
```

**Data Analysis**

```python
# statistical meassure of the dataset

df.describe()
```

```python
# checking the distribution of heart disease status

df["target"].value_counts()
```

target label representation:
- 1 = present of heart disease
- 0 = absent of heart disease

Splitting of data into features and target 

```python
# store features and target in X and Y respectively

X = df.drop(columns = ["target"])
Y = df["target"]
```

**Model Development**

```python
# splitting the dataset into training and test data

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42, stratify = Y)
```

```python
# create the logistic model

model = LogisticRegression(max_iter = 1000, random_state = 42)

# train the model with training data

model.fit(X_train, Y_train)
```

**Model Evaluation**

```python
# compare the prediction of the model with training data
train_y_pred = model.predict(X_train)

train_df = pd.DataFrame({
    "Actual": Y_train,
    "Prediction": train_y_pred
})

train_df.head()
```

```python
# compare the prediction of the model with test data
test_y_pred = model.predict(X_test)

test_df = pd.DataFrame({
    "Actual": Y_test,
    "Prediction": test_y_pred
})

test_df.head()
```

```python
# Evaluate the model performance with classification report

train_rpt = classification_report(Y_train, train_y_pred)
test_rpt = classification_report(Y_test, test_y_pred)

print(f"Training Data classification report is \n {train_rpt}")
print(f"Test Data classification report is \n {test_rpt}")
```
